```python
# Palindrom
num = int(input("Enter a number: "))

original = num
rev = 0

while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10
    print(digit)

if original == rev:
    print("Number is Palindrome")
else:
    print("Number is not Palindrome")
    
    
    
```

```python
class Time:
    def __init__(self, hrs, min, sec):
        self.hrs = hrs
        self.min = min
        self.sec = sec

    def __add__(self, other):
        s = self.sec + other.sec
        m = self.min + other.min
        h = self.hrs + other.hrs

        if s >= 60:
            m += s // 60
            s = s % 60

        if m >= 60:
            h += m // 60
            m = m % 60

        return Time(h, m, s)

    def display(self):
        print(self.hrs, "Hours", self.min, "Minutes", self.sec, "Seconds")


t1 = Time(2, 55, 50)
t2 = Time(1, 30, 30)

t3 = t1 + t2

t3.display()
```

```python
class Student:
    def __init__(self, fullname, marks):
        self.name = fullname
        self.marks = marks
        print("Constructor is called")

s1 = Student("Prince", 88)
print(s1.name,s1.marks )

s2 = Student("Karan", 92)
print(s2.name, s2.marks)
```

```python
class Student:
    # default constructor
    def __init__(self):
        pass
    
    # parameterized constructor
    def __init__(self, fullname, marks):
        self.name = fullname
        self.marks = marks
        print("Constructor is called")
```

```python
class Student:
    college_name = "ABC College"
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    
    def welcome(self):
        print("Welcome student,", self.name)
        
    def get_marks(self):
        return self.marks

s1 = Student("karan",97)
s1.welcome()
print(s1.get_marks())
```

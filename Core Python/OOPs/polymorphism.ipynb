```python
class Employee:
    def __init__(self, id, name, salary):
        self.id = id
        self.name = name
        self.salary = salary

    def getId(self):
        return self.id

    def setId(self, newId):
        self.id = newId

    def getName(self):
        return self.name

    def setName(self, newName):
        self.name = newName

    def getSalary(self):
        return self.salary

    def setSalary(self, newSalary):
        self.salary = newSalary

    def display(self):
        print(f"Id={self.id} Name={self.name} Salary={self.salary}")
        
    def calSalary(self):
        print("Final salary of employee=",self.salary)

class Hr(Employee):
    def __init__(self, id,name,salary,com):
        super().__init__(id,name,salary)
        self.com = com
        
    def getCom(self):
        return self.com
    def setCom(self, newCom):
        self.com = newCom
    def display(self):
        super().display()
        print(f"Com={self.com}")
    def calSalary(self):
        #return super().calSalary() + self.com
        print(f"sal of Hr={self.com+self.Salary}")
        
class Dev(Employee):
    def __init__(self, id,name,salary,bonus):
        super().__init__(id,name,salary)
        self.bonus = bonus
        
    def getBonus(self):
        return self.bonus
    def setBonus(self, newBonus):
        self.bonus = newBonus
    def display(self):
        super().display()
        print(f"Bonus={self.bonus}")
    def calSalary(self):
        #return super().calSalary() + self.bonus
        print(f"sal of Dev={self.bonus+self.Salary}")
    
e = Employee(101, "Sudhir", 50000)
d = Dev(102, "Ram", 60000, 5000)
h = Hr(103, "Sham", 70000,4000)

# e.display()
# d.display()
# h.display()

e.calSalary()
d.calSalary()
h.calSalary()

```

```python
#1
def demo():
    yield 10
    print("First")
    yield 20
    print("Second")
    yield 30
    print("Third")
d=demo()
#print(d.__next__())
print(next(d))
print(next(d))
print(next(d))
print(next(d))
```

```python
#2
def demo():
    for i in range(1,11):
        yield i
d = demo()
print(next(d))
print(next(d))
print(next(d))
print(next(d)) # here next() is a function which is used to get the next value from the generator object. It will raise StopIteration exception when there are no more items to return.
```

```python
#3
import sys
li = [i for i in range(1,1000000)]
l2 = (i for i in range(1,1000000))
print(sys.getsizeof(li))
print(sys.getsizeof(l2)) # here we are using generator expression which is more memory efficient than list comprehension. It will not store all the values in memory at once. It will generate the values on the fly when we iterate over it.
```

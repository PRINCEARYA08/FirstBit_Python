```python
#1
def demo():
    yield 10
    print("First")
    yield 20
    print("Second")
    yield 30
    print("Third")
d=demo()
#print(d.__next__())
print(next(d))
print(next(d))
print(next(d))
print(next(d))
```

```python
#2
def demo():
    for i in range(1,11):
        yield i
d = demo()
print(next(d))
print(next(d))
print(next(d))
print(next(d))
```
